# Notebook 01 – Setup

End-to-end setup for the complaint-router project: create an isolated virtual environment, install all dependencies, register a dedicated Jupyter kernel, and launch the vLLM inference server.

---
## Step 1 – Create the virtual environment

Creates a self-contained Python virtual environment in the `.venv/` directory at the project root. All subsequent installs are isolated from the system Python, making the setup fully reproducible and easy to tear down.

```python
%cd .. 
!python3 -m venv .venv
```

## Step 2 – Install the project and dev dependencies

Installs the `complaint_router` package in *editable* mode (`-e`) from the local source tree, plus every optional dependency listed under `[dev]` in `pyproject.toml` (pytest, ruff, mypy, etc.). Editable mode means live code changes are reflected immediately without reinstalling.

```python
%cd ..
!.venv/bin/python -m pip install -e ".[dev]"
```

## Step 3 – Install and register the Jupyter kernel

Installs `ipykernel` into the venv, then registers it with Jupyter under the name `complaint-router`. After running this step, select **"Python (complaint-router .venv)"** from the kernel picker at the top-right of any notebook in this project to ensure all cells run inside the isolated environment.

```python
%cd .. 
!.venv/bin/python -m pip install ipykernel
!.venv/bin/python -m ipykernel install --user --name complaint-router --display-name "Python (complaint-router .venv)"
```

## Step 4 – Serve the model with vLLM

Starts the vLLM OpenAI-compatible inference server for `Mistral-7B-Instruct-v0.3` on port **8000**.


> **Note:** Run this command in a **separate terminal** (not inside the notebook). The process blocks while the server is running. The agent code in cells below targets `http://localhost:8000`.

```python
VLLM_USE_TRITON_FLASH_ATTN=0 vllm serve mistralai/Mistral-7B-Instruct-v0.3 \
   --served-model-name mistralai/Mistral-7B-Instruct-v0.3 \
   --api-key abc-123 --port 8000
```
